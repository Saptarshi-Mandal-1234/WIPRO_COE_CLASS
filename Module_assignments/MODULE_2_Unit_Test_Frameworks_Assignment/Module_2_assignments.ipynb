# Assignment 7: Page Object Model (POM) Restructure


## Login Page


```python
from selenium.webdriver.common.by import By

class LoginPage:

    username = (By.ID, "user-name")
    password = (By.ID, "password")
    login_button = (By.ID, "login-button")

    def __init__(self, driver):
        self.driver = driver

    def login(self, username, password):
        self.driver.find_element(*self.username).send_keys(username)
        self.driver.find_element(*self.password).send_keys(password)
        self.driver.find_element(*self.login_button).click()
```

## Dashboard Page


```python
from selenium.webdriver.common.by import By

class DashboardPage:

    products_title = (By.CLASS_NAME, "title")

    def __init__(self, driver):
        self.driver = driver

    def get_title(self):
        return self.driver.find_element(*self.products_title).text
```

## Test Case


```python
from selenium import webdriver
import time

driver = webdriver.Chrome()

driver.get("https://www.saucedemo.com/")
driver.maximize_window()

time.sleep(2)

login_page = LoginPage(driver)
dashboard_page = DashboardPage(driver)

time.sleep(2)

login_page.login("standard_user", "secret_sauce")

time.sleep(3)

assert dashboard_page.get_title() == "Products"

print("Login test passed successfully.")

time.sleep(5)

driver.quit()
```

## Executing the Test Cases


```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time

data = pd.read_csv("login_data.csv")

for index, row in data.iterrows():

    driver = webdriver.Chrome()

    driver.get("https://www.saucedemo.com/")
    driver.maximize_window()

    time.sleep(2)

    driver.find_element(
        By.ID, "user-name"
    ).send_keys(row["username"])

    time.sleep(1)

    driver.find_element(
        By.ID, "password"
    ).send_keys(row["password"])

    time.sleep(1)

    driver.find_element(
        By.ID, "login-button"
    ).click()

    time.sleep(3)

    if row["expected"] == "success":

        title = driver.find_element(
            By.CLASS_NAME, "title"
        ).text

        assert title == "Products"

        print(
            f"Test {index + 1}: Login successful"
        )

    else:

        error = driver.find_element(
            By.CSS_SELECTOR,
            "h3[data-test='error']"
        ).text

        assert error != ""

        print(
            f"Test {index + 1}: Validation error displayed"
        )

    time.sleep(3)

    driver.quit()

    time.sleep(2)
```

## Capturing Screenshots for Failed Tests


```python
import pytest
from selenium import webdriver

@pytest.hookimpl(hookwrapper=True)
def pytest_runtest_makereport(item, call):

    outcome = yield
    report = outcome.get_result()

    setattr(item, "rep_" + report.when, report)


@pytest.fixture
def driver(request):

    driver = webdriver.Chrome()
    driver.maximize_window()

    yield driver

    if request.node.rep_call.failed:

        screenshot = f"{request.node.name}.png"

        driver.save_screenshot(screenshot)

        pytest_html = request.config.pluginmanager.getplugin("html")

        if pytest_html:

            extra = getattr(request.node, "extras", [])

            extra.append(
                pytest_html.extras.image(screenshot)
            )

            request.node.extras = extra

    driver.quit()
```
